```python
import pandas as pd
import json

# 🔹 Load the CSV file (make sure it's in the same directory as this script)
file_path = "statements.csv"  # Update with the correct file path if needed
df = pd.read_csv(file_path)

# 🔹 Convert to JSON format
statements_json = []
for _, row in df.iterrows():
    statement_id = f"{row['Category']}.{row['Type']}.{row['Number']}"
    statements_json.append({"id": statement_id, "statement": row["Statement"]})

# 🔹 Save JSON file
json_file_path = "statements.json"
with open(json_file_path, "w") as f:
    json.dump(statements_json, f, indent=4)

print(f"✅ JSON file saved as {json_file_path}")
```
